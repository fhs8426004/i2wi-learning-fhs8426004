```python
def check_test_value():
    try:
        value = float(input("検査値を入力してください: "))

        # 判定処理（正常 -> 異常 -> 危険の順に評価）
        if 10 <= value <= 20:
            print(f"結果: {value} は【正常】です。")
        elif (5 <= value < 10) or (20 < value <= 25):
            print(f"結果: {value} は【異常】です。")
        else:
            print(f"結果: {value} は【危険】です。")

    except ValueError:
        print("エラー: 数値を入力してください。")

# プログラムの実行
check_test_value()
```

```python
# --- 基準値の設定（ここを変更するだけで範囲をカスタマイズ可能） ---
NORMAL_MIN, NORMAL_MAX = 10.0, 20.0  # 正常範囲
CRITICAL_MIN, CRITICAL_MAX = 5.0, 25.0  # 危険範囲の境界

def evaluate_test_value(value: float) -> str:
    """検査値を判定してメッセージを返す関数"""
    if NORMAL_MIN <= value <= NORMAL_MAX:
        return "【正常】"

    # 低い側の判定
    if CRITICAL_MIN <= value < NORMAL_MIN:
        return "【異常】基準値より低いです"
    if value < CRITICAL_MIN:
        return "【危険】極めて低いです"

    # 高い側の判定
    if NORMAL_MAX < value <= CRITICAL_MAX:
        return "【異常】基準値より高いです"
    return "【危険】極めて高いです"


def main():
    print("=== 検査値判定プログラム ===")
    print("（'q' と入力すると終了します）\n")

    while True:
        user_input = input("検査値を入力してください: ").strip()

        # 終了処理
        if user_input.lower() == 'q':
            print("プログラムを終了します。")
            break

        try:
            val = float(user_input)
            result = evaluate_test_value(val)
            print(f"-> 測定値: {val} | 判定: {result}\n")
        except ValueError:
            print("エラー: 数値または 'q' を入力してください。\n")

if __name__ == "__main__":
    main()
```
