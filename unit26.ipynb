```python
def spec_text(name, color, wheels, gears, tire_inch):
  lines = [
      f"名前     : {name}",
      f"色       : {color}",
      f"車輪     : {wheels}本",
      f"ギア     : {gears}段"
      f"タイヤ径 :{tire_inch}インチ"
  ]
  return "\n".join(lines)

print(spec_text("通勤号", "青", 2, 7, 26.0))
```

```python
%%writefile bicycle.py
class Wheel:
    """概念: 車輪（自転車の部品）"""

    def __init__(self, inch):
        # 公理: タイヤ径は0より大きい
        if inch <= 0:
            raise ValueError(f"タイヤ径は0より大きい数です: {inch}")
        self.inch = inch


class Bicycle:
    """概念: 自転車"""

    def __init__(self, name, color, gears, wheels):
        # 公理: 車輪は2つ以上、ギアは1段以上
        if len(wheels) < 2:
            raise ValueError(f"車輪は2つ以上です: {len(wheels)}")
        if gears < 1:
            raise ValueError(f"ギアは1段以上です: {gears}")
        self.name = name
        self.color = color
        self.gears = gears
        self.wheels = wheels  # part-of: 車輪を部品として持つ
        self.gear = 1

    def shift_up(self):
        if self.gear == self.gears:
            print("これ以上上がりません")
            return
        self.gear = self.gear + 1

    def shift_down(self):
        if self.gear == 1:
            print("これ以上下がりません")
            return
        self.gear = self.gear - 1

    def to_spec(self):
        return {
            "name": self.name,
            "color": self.color,
            "gears": self.gears,
            "wheels": [w.inch for w in self.wheels],
            "concepts": ["自転車", "車輪"],
            "relations": ["車輪 part-of 自転車"],
            "axioms": ["車輪は2つ以上", "ギアは1段以上", "タイヤ径は0より大きい", "いまのギアは1〜ギア段数"],
        }


class ElectricBicycle(Bicycle):
    """概念: 電動アシスト自転車（is-a 自転車）"""

    def __init__(self, name, color, gears, wheels, battery_wh):
        super().__init__(name, color, gears, wheels)
        # 公理: 電池の容量は0より大きい
        if battery_wh <= 0:
            raise ValueError(f"電池の容量は0より大きい数です: {battery_wh}")
        self.battery_wh = battery_wh

    def to_spec(self):
        spec = super().to_spec()
        spec["battery_wh"] = self.battery_wh
        spec["concepts"] = spec["concepts"] + ["電動アシスト自転車"]
        spec["relations"] = spec["relations"] + ["電動アシスト自転車 is-a 自転車"]
        spec["axioms"] = spec["axioms"] + ["電池の容量は0より大きい"]
        return spec


if __name__ == "__main__":
    bike = Bicycle("通勤号", "青", 7, [Wheel(26.0), Wheel(26.0)])
    print(bike.to_spec())
```

```python
%%writefile bicycle.py

class Bicycle:
    def __init__(self, name, color, gears):
        self.name = name
        self.color = color
        self.gears = gears
        self.gear = 1

    def shift_up(self):
        if self.gear == self.gears:
            print("これ以上上がりません")
            return
        self.gear = self.gear + 1

    def shift_down(self):
        if self.gear == 1:
            print("これ以上下がりません")
            return
        self.gear = self.gear - 1

    def spec(self):
        return f"{self.name}（{self.color}）ギア {self.gear}/{self.gears} 段"
```
